conditinal logic

```python
import numpy as np

spending = np.array([500, 2000, 7000, 12000])

# Apply business rule
discount = np.where(
    spending < 1000, 0,
    np.where(spending < 5000, 5, 10)
)

print(discount)
```

Multi-Condition Filtering

```python
customers = np.array([100, 2000, 6000, 15000])

# High-value customers (5000–15000)
high_value = customers[(customers >= 5000) & (customers <= 15000)]

print(high_value)
```

Data Cleaning (Handling Missing Values)

```python
data = np.array([10, 20, np.nan, 40, np.nan])

# Replace NaN with mean
mean = np.nanmean(data)
cleaned = np.where(np.isnan(data), mean, data)

print(cleaned)
```

Aggregations

```python
sales = np.array([100, 200, 300, 400, 500])

print("Total:", np.sum(sales))
print("Average:", np.mean(sales))
print("Max:", np.max(sales))
print("Growth %:", (sales[-1] - sales[0]) / sales[0] * 100)
```

Group-like Operations

```python
region = np.array(['A', 'B', 'A', 'B'])
revenue = np.array([100, 200, 300, 400])

# Sum revenue for region A
print(np.sum(revenue[region == 'A']))
```

Broadcasting (scaling Data)

```python
prices = np.array([100, 200, 300])
tax_rate = 0.13

final_price = prices + prices * tax_rate
print(final_price)
```

Matrix Operations

```python
# Product sales matrix
sales = np.array([[10, 20],
                  [30, 40]])

# Multiply with price vector
price = np.array([5, 10])

revenue = sales @ price
print(revenue)
```

Outliers

```python
data = np.array([10, 12, 11, 500, 13, 12])

mean = np.mean(data)
std = np.std(data)

outliers = data[np.abs(data - mean) > 2 * std]
print(outliers)
```
