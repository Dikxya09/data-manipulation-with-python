Filtering

```python
a = np.array([10,20,30,40])
filtered = a[a > 20]
print(filtered)
```

Filter with mask

```python
a = np.array([10,20,30,40])
mask = a > 25
print("Number greater than 25 is", a[mask])
```

array stacking & splitting

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.vstack((a,b))) #Vertical stack
print(np.hstack((a,b))) #Horizontal stack
```

```python
c = np.array([[1,2],[3,4]])
print(c)
```

```python
print(np.hsplit(c, 2)) #Horizontal split
```

```python
split = np.hsplit(c, 2)
for s in split:
    print(s)
```

```python
vsplit = np.vsplit(c, 2)
for s in vsplit:
    print(s)
```
