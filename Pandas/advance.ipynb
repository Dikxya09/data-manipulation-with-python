#percentile:

percentiles are used to understand the distribution of data and to identify outliers. They indicate the value below which a given percentage of observations in a dataset falls.

```python
import numpy as np 

data = np.array([1,2,3,4,5])
p25 = np.percentile(data, 25)
p75 = np.percentile(data, 75)
p50 = np.percentile(data, 50)
print("25th Percentile:", p25)
print("75th Percentile:", p75)
print("50th Percentile:", p50)
```

Outliers:to find the outliers:
IQR = q3 - q1 
lower_bound = q1 - 1.5 * IQR
upper_bound = q3 + 1.5 * IQR

```python
import pandas as pd
data = [10, 12, 14, 15, 18, 20, 22, 25, 30, 100]
df = pd.DataFrame(data, columns=["Values"])
print(df)
```

```python
#IQR method to detect outliers
# Step 1: Calculate Q1 and Q3
Q1 = df['Values'].quantile(0.25)
Q3 = df['Values'].quantile(0.75)
print("Q1:", Q1)
print("Q3:", Q3)
```

```python
lower_bound = Q1 - 1.5 * (Q3 - Q1)
upper_bound = Q3 + 1.5 * (Q3 - Q1)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[(df["Values"] < lower_bound) | (df["Values"] > upper_bound)]
print("Outliers:",outliers)
```

Z-score method:standard score to find the outliers
z = x-mean / sd 
0 = exactly mean
+1 = above mean
-1 = below mean
+3 or -3 =outlier 

```python
import pandas as  pd 
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)
print(df.head(10))
print(df.tail(10))
print("*---")
print(df.info())
print(df.describe())
```

```python
import pandas as pd

# Load dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

# Step 1: Calculate Q1 and Q3
Q1 = df["petal_length"].quantile(0.25)
Q3 = df["petal_length"].quantile(0.75)

# Step 2: Calculate IQR
IQR = Q3 - Q1

# Step 3: Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Step 4: Extract outliers
outliers = df[(df["petal_length"] < lower_bound) | (df["petal_length"] > upper_bound)]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:\n", outliers)
```

```python
import numpy as np

mean = df["sepal_length"].mean()
std = df["sepal_length"].std()

df["z_score"] = (df["sepal_length"] - mean) / std 
outliers = df[np.abs(df["z_score"]) > 3]

print("Outliers:\n", outliers)
```

```python
import pandas as pd
import numpy as np
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

df['petal_sepal_ratio']=df['petal_length']/df['sepal_length']

mean_ratio = df['petal_sepal_ratio'].mean()
std_ratio = df['petal_sepal_ratio'].std()

df['ratio_zscore'] = (df['petal_sepal_ratio'] - mean_ratio) / std_ratio
outliers=df[np.abs(df['ratio_zscore'])>3]
print('OUTLIERS',outliers)
```
