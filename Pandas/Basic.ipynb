```python
import pandas as pd
data = {
    "Name": ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    "Age": [25, 30, 35, 40, 45],
    "City": ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    "Salary": [70000, 80000, 90000, 100000, 110000]
}
df = pd.DataFrame(data)
print("DataFrame created from the dictionary:")
print(df)
```

```python
column = df['Age']  # Accessing the 'Age' column
df.to_csv('output.csv' , index=False) # This line saves the DataFrame to a CSV file named 'output.csv' without including the index column.
```

```python
print("Descriptive statistics of the DataFrame:")
print(df.describe())
```

```python
print(f'Shape of the DataFrame: {df.shape}')
print(f'Columns in the DataFrame: {df.columns}')
```

```python
print("Name (single column):")
print(df["Name"])
```

```python
#selecting multiple columns
print("Name and City (multiple columns):")
subset = df[["Name", "City"]]
print(subset)
```

```python
#High Salary
high_salary = df[df["Salary"] > 90000]
print(high_salary)
```

```python
#filtering rows based on multiple conditions
filtered_data = df[(df["Age"] > 30) & (df["Salary"] > 90000)]
print("Filtered data (Age > 30 and Salary > 90000):")
print(filtered_data)
```

```python
#adding new column
df["Bonus"] = df["Salary"] * 0.1  # Adding a new column 'Bonus' which is 10% of the Salary
print(df)
```

```python
#inserting a new column at a specific position
df.insert(2, "Department", ["HR", "IT", "Finance", "Marketing", "Sales"])  # Inserting 'Department' column at index 2
print(df)
```

```python
#updating values in a column
df.loc[0, "Salary"] = 75000  # Updating the salary of the first row to 75000    
print(df)
```

```python
#increasing salary by 5% for all employees
df["Salary"] = df["Salary"] * 1.05  # Increasing salary by 5% for all employees
print(df)
```

```python
#Selecting Subsets of Rows and Columns
df.loc[[0,1]][["City","Salary"]]
```
